# Import libraries

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

# Import Dataset

```python
# Reading the dataset file for processing and model building.
dataset = pd.read_csv('../../datasets/Position_Salaries.csv')
# print(dataset)

# Separating the columns for variables x and y
x = dataset.iloc[:, 1:2].values
y = dataset.iloc[:, 2].values

# print(x)
# print(y)
```

# Training Model

```python
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures

poly_regressor = PolynomialFeatures(degree=4)
x_poly = poly_regressor.fit_transform(x)

linear_regressor = LinearRegression()
linear_regressor.fit(x_poly, y)

```

# Visualising

```python
# Visualising 
plt.scatter(x, y, color = 'red')
plt.plot(x, linear_regressor.predict(poly_regressor.fit_transform(x)), color = 'blue')
plt.title('Truth or Bluff (Polynomial Regression)')
plt.xlabel('Position level')
plt.ylabel('Salary')
plt.show()
```
